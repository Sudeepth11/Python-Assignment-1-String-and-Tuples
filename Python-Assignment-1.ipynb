1.String Concatenation

```python
str1="Hello"
str2=input("Enter your Name: ")
halfstr=str1+" "+str2
print(halfstr)
str3=", welcome to Python programming"
fullstr=halfstr+str3
print(fullstr)
```

2. String Slicing and Indexing

```python
print(fullstr[0])
print(fullstr[-1])
print(fullstr[:5])
print(fullstr[-11:])
print(fullstr[::-1])
py_strt=fullstr.find("Python")
py_end=py_strt+len("Python")
print(fullstr[py_strt:py_end])
```

3. String Methods

```python
strM="Python beginner tutorial"
print(strM.upper())
print(strM.lower())
print(strM.capitalize())
print(strM.count('t'))
print(strM.replace('Python','Machine Learning'))
```

Tuples (Creation, Modification and Access)
